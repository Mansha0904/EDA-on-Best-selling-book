```python
data_23['Author']=data_23['Author'].fillna('No author Specified') 


filtered_author=pd.DataFrame() 
filtered_author=data_23['Author'].str.split(',',expand=True).stack() 
filtered_author=filtered_author.to_frame()
filtered_author.columns=['Author']
Author=filtered_author.groupby(['Author']).size().reset_index(name='Total Content')
Author=Author[Author.Author !='No author Specified'] 
Author=Author.sort_values(by=['Total Content'],ascending=False) 
top_author=Author.head(10) 
top_author=top_author.sort_values(by=['Total Content'])
x = top_author["Author"]
y = top_author["Total Content"]

# PLOT the Bar Graph
plt.style.use("fivethirtyeight")
plt.figure(figsize = (20,10))
plt.title('Top 10 Author')
sns.barplot(x = x, y = y)
plt.show()
```
